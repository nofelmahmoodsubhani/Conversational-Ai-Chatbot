```python
#SETTING UP THE MODEL
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage

model = init_chat_model("gpt-4o-mini", model_provider="openai")
```

```python
#.invoke is used to pass structured input
#.content extracts only the content filters meta data
```

```python
#SIMPLE SYNTAX FOR MESSAGE PASSING
messages = [
    SystemMessage("Translate the following from English into Italian"),
    HumanMessage("my name is nofel!")
]

response = model.invoke(messages)
print(response.content)
```

```python
#OTHER FORMATS FOR INVOKE
response1 = model.invoke("Hello")
print(response1.content)

response2 = model.invoke([{"role": "user", "content": "Hello"}])
print(response2.content)

response3 = model.invoke([HumanMessage("Hello")])
print(response3.content)
```

```python
#token is just a variable for storing each iteration of comuputation 
#.stream used when instead of waiting for the full reply, it yields tokens as they arrive
```

```python
#STREAMING THE OUTPUT/TOKEN
for token in model.stream(messages):
    print(token.content, end=" ")
```

```python
#ChatPromptTemplate used to build structured prompts
#.from_messages used to actually create the temelate
```

```python
#TEMPELATES
from langchain_core.prompts import ChatPromptTemplate

system_template = "Translate the following from English into {language}"

prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_template), 
     ("user", "{text}")
    ])

prompt = prompt_template.invoke({"language": "Italian", "text": "hi!"}) #FILLING of tempelate with invoke
prompt
```

```python
prompt.to_messages()  #CONVERSION INTO LIST MODEL UNDERSTANDS
```

```python
response = model.invoke(prompt)
print(response.content)
```
